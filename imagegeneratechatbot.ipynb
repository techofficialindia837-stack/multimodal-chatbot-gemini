```python
from PIL import Image
import google.generativeai as genai

genai.configure(api_key="AIzaSyClLleub1jtMqCPPv0yf5FbZ3AGeX_wX3U")

model = genai.GenerativeModel("gemini-2.5-flash")

def multimodal_chat(prompt, image_path=None):

    if image_path:
        img = Image.open(image_path)
        response = model.generate_content([prompt, img])
    else:
        response = model.generate_content(prompt)

    return response.text


# Test
print(multimodal_chat("What is AI?"))
```

```python
while True:

    question = input("Ask: ")

    if question == "exit":
        break

    image_path = input("Enter image path (or press enter): ")

    if image_path == "":
        print(multimodal_chat(question))
    else:
        print(multimodal_chat(question, image_path))
```
